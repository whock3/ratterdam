### Making supplementary figures for shuffled GLMs 
#### 3 figs: CD shuff, CD native shuff, time shuff

```python
from matplotlib import pyplot as plt
import numpy as np, pandas as pd, seaborn as sns
```

```python
%matplotlib qt5
%qtconsole
```

```python
data_dir = 'E:\\Ratterdam\\repetition_manuscript\\Supplementary_Figures\\GLM_shuffles\\'
```

```python
df = pd.read_csv(data_dir + "time_shuff.csv")
```

```python
real_time = 0.28
real_CD = 0.35
```

```python
# time
df = pd.read_csv(data_dir + "time_shuff.csv")
ax = sns.histplot(data=df, x = 'x', color = 'firebrick')
ax.vlines(real_time,0,350,'r')
ax.set_xlabel("Proportion of Oriented Fields", fontsize=16)
ax.set_ylabel("Frequency (Permutation Shuffle)", fontsize=16)

plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.set_title("Proportion of Fields in Population with Significant \nFiring Rate Non-stationarity",fontsize=16)
```

```python
# CD shuffle permutation 
df = pd.read_csv(data_dir + "CD_shuff.csv")
ax = sns.histplot(data=df, x = 'x', color = 'cornflowerblue')
ax.vlines(real_CD,0,150,'b')
ax.set_xlabel("Proportion of Oriented Fields", fontsize=16)
ax.set_ylabel("Frequency (Permutation Shuffle)", fontsize=16)

plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.set_title("Proportion of Fields in Population with \nTuning for Current Direction",fontsize=16)
```

```python
# CD naive permutation 
df = pd.read_csv(data_dir + "CD_naive.csv")
ax = sns.histplot(data=df, x = 'x', color = 'navy')
ax.vlines(real_CD,0,250,'b')
ax.set_xlabel("Proportion of Oriented Fields", fontsize=16)
ax.set_ylabel("Frequency (Naive Shuffle)", fontsize=16)

plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.set_title("Proportion of Fields in Population with \nTuning for Current Direction",fontsize=16)
```
