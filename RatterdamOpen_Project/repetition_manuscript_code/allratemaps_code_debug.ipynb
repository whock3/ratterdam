```python

path = 'E:\\Ratterdam\\R_data_repetition\\' # Your path to data 
with open(path + "20220405-124315_superPopulationRepetition.pickle","rb") as f:
    superpop = pickle.load(f)  
```

```python
with open('E:\\Ratterdam\\repetition_manuscript\\Supplementary_Figures\\interneurons\\20240628_INlist.json') as f:
    inlist = json.load(f)
```

```python
def checkIN(rat,day,name):
    isIN = False
    for cell in inlist:
        if cell['rat'] == rat and cell['day'] == day and cell['clust'] == name:
            isIN = True
    return isIN
```

```python
cmap = util.makeCustomColormap()
r,c = 10,5
fig, ax = plt.subplots(r,c,figsize=(20,20))
i=0
j=0
for rat in superpop.keys():
    for day in superpop[rat].keys():
        for unit in superpop[rat][day]['units'].values():

            if not checkIN(rat, day, unit.name):


                if i%(r*c) == 0:

                    for _ax in fig.axes:
                        _ax.set_xticks([])
                        _ax.set_yticks([])
                        for spineloc in ['top','bottom','left','right']:
                            _ax.spines[spineloc].set_visible(False)

                    fig, ax = plt.subplots(r,c,figsize=(20,20))
                    fig.subplots_adjust(wspace=0, hspace=0.2)

                    j=0


                ax = fig.axes[j]
                ax.imshow(unit.repUnit.rateMap2D, origin='lower', aspect='auto', interpolation='None', 
                        cmap=cmap, vmax=np.nanpercentile(unit.repUnit.rateMap2D, 98),
                    extent=[wmDef.xedges[0], wmDef.xedges[-1], wmDef.yedges[0], wmDef.yedges[-1]])

                t = unit.name.split("\\")
                t2 = t[1].split('.')
                if len(unit.fields) > 0:
                    color = 'black'
                    ax.set_title(f"{rat}{day} {t[0]} cell {t2[1]}",fontsize=8,y=0.9,color=color)

                if len(unit.fields) == 0 or (rat == "R808" and day == "D7" and unit.name == "TT12\\cl-maze1.8"):
                    color = 'red'
                    for entry in excludeList:
                        entry = entry.split(",")
                        if entry[0] == rat and entry[1] == day and entry[2] == t[0] and entry[3] == t2[1]:
                            print("match found")
                            code = entry[4]
                    ax.set_title(f"{rat}{day} {t[0]} cell {t2[1]} - exclusion code = {code}",fontsize=8,y=0.9,color=color)

      
                for spineloc in ['top','bottom','left','right']:
                    ax.spines[spineloc].set_visible(False)
                ax.set_xticks([])
                ax.set_yticks([])
                ax.set_aspect('equal')
            
                i += 1
                j += 1
```
